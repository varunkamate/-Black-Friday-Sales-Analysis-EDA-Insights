```python
#Calculate Q1, Q3, and IQR for Purchase_Amount.

q1=df["Purchase_Amount"].quantile(0.25)
q3=df["Purchase_Amount"].quantile(0.75)

IQR=q3-q1


print(q1,q3,IQR)
```

```python
#Identify outliers using the IQR rule.

lower_bound=q1-IQR*1.5
upper_bound=q3+IQR*1.5

outliers=df[(df["Purchase_Amount"]<lower_bound)|(df["Purchase_Amount"]>upper_bound)]
print(outliers)
```

```python
outliers_count = outliers.shape[0]
outliers_count
```

```python
#Compute Z-scores for Purchase_Amount.
#Which values exceed ±3?
#Compare IQR vs Z-score results. Do they flag the same outliers?

data=df[["Customer_ID","Purchase_Amount"]].dropna()

from scipy.stats import zscore

data["zscore"]=zscore(data["Purchase_Amount"])
zoutliers=data[abs(data["zscore"])>3]

print(data[["Customer_ID","Purchase_Amount","zscore"]])
print("zscore outliers is:==>")
print(zoutliers)
```
